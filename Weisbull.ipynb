```python
import numpy as np
import matplotlib.pyplot as plt
import math

# Parámetros de la distribución de Weibull
beta = 2.5  # Parámetro de forma
eta = 8000  # Parámetro de escala (vida característica)

# Fórmula de la Función de Fallo Acumulado (CDF) de Weibull
def weibull_cdf(t, beta, eta):
  """
  Calcula la probabilidad de fallo acumulado en el tiempo t.
  """
  return 1 - np.exp(-(t / eta)**beta)

# Generar puntos de tiempo
tiempo = np.linspace(0, 2 * eta, 400)  # Generamos puntos hasta el doble de la vida característica

# Calcular la CDF para cada punto de tiempo
cdf_valores = weibull_cdf(tiempo, beta, eta)

# Fórmula de la Función de Densidad de Probabilidad (PDF) de Weibull (opcional para la gráfica)
def weibull_pdf(t, beta, eta):
  """
  Calcula la función de densidad de probabilidad en el tiempo t.
  """
  return (beta / eta) * (t / eta)**(beta - 1) * np.exp(-(t / eta)**beta)

pdf_valores = weibull_pdf(tiempo, beta, eta)

# Crear la gráfica de la Función de Fallo Acumulado (CDF)
plt.figure(figsize=(10, 6))
plt.plot(tiempo, cdf_valores, label=f'CDF (β={beta}, η={eta})', color='blue')

# Marcar la vida característica (eta)
plt.axvline(eta, color='red', linestyle='--', label=f'Vida Característica (η = {eta:.0f} horas)')
plt.scatter(eta, weibull_cdf(eta, beta, eta), color='red', marker='o')
plt.annotate(f'P(Fallo en {eta:.0f}h) ≈ {weibull_cdf(eta, beta, eta):.2f}',
             xy=(eta, weibull_cdf(eta, beta, eta)),
             xytext=(eta + 500, weibull_cdf(eta, beta, eta) + 0.05),
             arrowprops=dict(facecolor='black', shrink=0.05))

# Etiquetas y título
plt.xlabel('Tiempo (horas)')
plt.ylabel('Probabilidad de Fallo Acumulado')
plt.title('Función de Fallo Acumulado (CDF) de Weibull')
plt.grid(True)
plt.legend()
plt.ylim(0, 1.05)
plt.xlim(0, tiempo[-1])
plt.tight_layout()
plt.show()

# Crear la gráfica de la Función de Densidad de Probabilidad (PDF) (opcional)
plt.figure(figsize=(10, 6))
plt.plot(tiempo, pdf_valores, label=f'PDF (β={beta}, η={eta})', color='green')
plt.xlabel('Tiempo (horas)')
plt.ylabel('Densidad de Probabilidad de Fallo')
plt.title('Función de Densidad de Probabilidad (PDF) de Weibull')
plt.grid(True)
plt.legend()
plt.xlim(0, tiempo[-1])
plt.tight_layout()
plt.show()
```
